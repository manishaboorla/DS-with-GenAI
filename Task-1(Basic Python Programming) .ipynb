#### 2.Python If-Else

```python
n=int(input().strip())

if n%2==1:
    print("Weird")
else:
    if 2<=n<=5:
        print("Not Weird")
    elif 6<=n<=20:
        print("Weird")
    else:
        print("Not Weird")
```

#### 6.Write a Function

```python
def is_leap(year):
    if(year%400==0):
        return True
    if(year%100==0):
        return False
    if(year%4==0):
        return True
    return False

year=int(input())
print(is_leap(year))
```
