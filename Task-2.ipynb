##### 1.Defanging an IP address

```python
address= input("enter an ip address")
class Solution:
    def defangIPaddr(self,address: str) -> str:
        result = []
        for char in address:
            if char == '.':
                result.append('[.]')
            else:
                result.append(char)
        return "".join(result)

a= Solution()
print(a.defangIPaddr(address))
```

##### 2.Find numbers with an even number of digits

```python
n= input("enter numbers:")
nums=[int(x) for x in n.split()]

def findNumbers(nums):
    even_count=0
    for i in nums:
        if len(str(i))%2==0:
            even_count+=1
    return even_count

print("Count of even-digits numbers:")
findNumbers(nums)
```

##### 3.Number of Good Pairs

```python
n = input("Enter numbers: ")
nums = [int(x) for x in n.split()]

def numIdenticalPairs(nums):
    pairs = 0
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] == nums[j]:
                pairs += 1
    return pairs

print("Total identical pairs:", numIdenticalPairs(nums))
```

##### 4.How many numbers are smaller than the current number

```python
n = input("Enter numbers: ")
nums = [int(x) for x in n.split()]

def smallerNumbersThanCurrent(nums):
    result =[]
    for i in range(len(nums)):
        count = 0
        for j in range(len(nums)):
            if nums[j]<nums[i]:
                count+=1
        result.append(count)
    return result

print("Resulting List:" ,smallerNumbersThanCurrent(nums))
        
    
```

##### 5.Subtract the product and sum of digits of an integer

```python
n = int(input("Enter a single integer: "))

def subtractProductAndSum(n):
    s = str(n)
    prod = 1
    summ = 0
    for digit in s:
        val = int(digit)
        prod *= val
        summ += val
        
    return prod - summ

print("Difference:", subtractProductAndSum(n))
```

##### 6.XOR opertion in an array

```python
n = int(input("Enter n: "))
start = int(input("Enter start: "))

def xorOperation(n, start):
    ans = 0
    for i in range(n):
        current = start + 2 * i
        ans = ans ^ current
    return ans

print("XOR Result:", xorOperation(n, start))
```
