##### 1. Running Sum of 1d Array

```python
class Solution:
    def runningSum(self, nums):
        total = 0
        answer = []

        for value in nums:
            total+=value
            answer.append(total)

        return answer
```

```python
Solution().runningSum([1,2,3,4])
```

```python
Solution().runningSum([3,1,2,10,1])
```

##### 2.Shuffle the Array

```python
class Solution:
    def shuffle(self, nums: list[int], n: int) -> list[int]:
        answer = []
        for i in range(n):
            answer.append(nums[i])
            answer.append(nums[i + n])
        return answer  
```

```python
Solution().shuffle([2,5,1,3,4,7],3)
```

##### 3.Kids With The Greatest Number of Candies

```python
from typing import List
class Solution:
    def kidsWithCandies(self, candies: List[int], extraCandies: int) -> List[bool]:
        max_candies=max(candies)
        answer= []
        
        for i in candies:
            answer.append(i+extraCandies>=max_candies)
        return answer
```

```python
Solution().kidsWithCandies([12,1,12],10)
```

```python
Solution().kidsWithCandies([4,2,1,1,2],1)
```
